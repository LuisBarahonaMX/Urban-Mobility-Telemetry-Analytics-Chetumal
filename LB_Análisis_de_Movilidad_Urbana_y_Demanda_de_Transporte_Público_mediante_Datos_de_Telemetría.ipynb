## Descripción del proyecto

Este notebook desarrolla un análisis exploratorio de datos de movilidad urbana utilizando registros operativos de unidades de transporte público de Chetumal, Quintana Roo.

El proyecto integra múltiples archivos CSV que contienen información temporal, identificadores de unidades, dispositivos de telemetría, coordenadas GPS y registros de ascenso y descenso de pasajeros.

Su objetivo es transformar estos datos en indicadores que permitan comprender el comportamiento de la demanda de transporte, identificar horarios de mayor movimiento y analizar patrones operativos por unidad, día y mes.

## Objetivos

- Integrar diferentes conjuntos de datos de transporte público.
- Realizar limpieza, normalización y validación de registros.
- Identificar coordenadas GPS inválidas y discontinuidades temporales.
- Calcular ascensos, descensos y movimiento total de pasajeros.
- Identificar horarios pico por unidad y a nivel general.
- Analizar la demanda mediante series temporales diarias y mensuales.
- Explorar indicadores de ocupación estimada utilizando contadores acumulados.
- Comparar patrones de movilidad mediante mapas de calor.
- Exportar indicadores consolidados para análisis posteriores.

## Metodología

**1. Integración y preparación de datos**

Se importan múltiples archivos CSV con Pandas, se unifican los registros y se estandarizan las variables numéricas, temporales y geográficas.

**2. Validación de telemetría**

Se identifican registros GPS inválidos, valores faltantes, incrementos negativos y discontinuidades temporales superiores a diez minutos entre registros consecutivos de un dispositivo.

**3. Análisis del movimiento de pasajeros**

Se calculan los ascensos y descensos registrados por los sensores de las puertas, generando indicadores de movimiento por dispositivo, día y hora.

**4. Análisis temporal**

Se construyen series temporales para identificar días de alta demanda, horarios pico y diferencias en el movimiento mensual.

**5. Visualización de patrones**

Se generan gráficas de líneas, barras y mapas de calor que permiten comparar el movimiento de pasajeros entre horas y meses. También se emplean promedios para facilitar la comparación de periodos con diferente cantidad de días registrados.

**6. Exportación de resultados**

Los indicadores procesados se exportan en archivos CSV para facilitar su consulta, reutilización e integración en futuros análisis.

## Tecnologías utilizadas

**Python | Pandas | NumPy | Matplotlib | Google Colab | CSV | Análisis de Series Temporales**

## Aplicaciones

Los indicadores obtenidos pueden servir como apoyo para estudiar horarios de demanda, identificar posibles irregularidades en los registros y desarrollar futuras herramientas de visualización y monitoreo operativo.

```python
import glob, os
import pandas as pd

files = sorted(glob.glob("/content/*.csv"))

dfs = []
for f in files:
    d = pd.read_csv(f, low_memory=False, dtype={"Unidad":"string", "Dispositivo":"string", "Empresa":"string"})
    d["source_file"] = os.path.basename(f)
    dfs.append(d)

df = pd.concat(dfs, ignore_index=True)
print(df.shape)
```

```python
import numpy as np

df = df.copy()

# Asegurar strings donde conviene
for c in ["Unidad","Dispositivo","Empresa","Ruta"]:
    if c in df.columns:
        df[c] = df[c].astype("string")

# Parsear datetime
df["FechaHora"] = pd.to_datetime(df["FechaHora"], errors="coerce")

# GPS inválido (0,0)
df["gps_valido"] = ~((df["Latitud"].fillna(0) == 0) & (df["Longitud"].fillna(0) == 0))

# Convertir a numérico
inc_cols = ["SubidasPuerta1","BajadasPuerta1","SubidasPuerta2","BajadasPuerta2"]
acc_cols = ["DatoSubidaPuerta1","DatoBajadaPuerta1","DatoSubidaPuerta2","DatoBajadaPuerta2"]
geo_cols = ["Latitud","Longitud"]

for c in inc_cols + acc_cols + geo_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")

# Limpiar incrementos negativos si existen
for c in inc_cols:
    df.loc[df[c] < 0, c] = np.nan

# Variables de tiempo
df["dia"] = df["FechaHora"].dt.date
df["hora"] = df["FechaHora"].dt.hour

print("FechaHora nulos:", df["FechaHora"].isna().sum())
print("GPS inválido:", (~df["gps_valido"]).sum(), "de", len(df))
print("Ruta vacía (NaN):", df["Ruta"].isna().mean() if "Ruta" in df.columns else "no existe")
```

```python
df = df.sort_values(["Dispositivo","FechaHora"]).reset_index(drop=True)

df["dt_seg"] = df.groupby("Dispositivo")["FechaHora"].diff().dt.total_seconds()
df["gap_10min"] = df["dt_seg"] > 600

print("Gaps >10min:", df["gap_10min"].sum())
df[["Dispositivo","FechaHora","dt_seg","gap_10min"]].head(15)
```

```python
df["subidas_total"] = df["SubidasPuerta1"].fillna(0) + df["SubidasPuerta2"].fillna(0)
df["bajadas_total"] = df["BajadasPuerta1"].fillna(0) + df["BajadasPuerta2"].fillna(0)
df["mov"] = df["subidas_total"] + df["bajadas_total"]

daily = (df.groupby(["Dispositivo","dia"], as_index=False)
           .agg(subidas=("subidas_total","sum"),
                bajadas=("bajadas_total","sum"),
                mov=("mov","sum"),
                registros=("ID_Trx","count"),
                gps_invalidos=("gps_valido", lambda s: (~s).sum()))
        )
daily["balance"] = daily["subidas"] - daily["bajadas"]
daily.sort_values(["dia","Dispositivo"]).head(10)
```

```python
by_hour = (df.groupby(["Dispositivo","dia","hora"], as_index=False)
             .agg(mov=("mov","sum"),
                  subidas=("subidas_total","sum"),
                  bajadas=("bajadas_total","sum"))
          )

peak_global = (by_hour.groupby("hora", as_index=False)["mov"].sum()
                     .sort_values("mov", ascending=False))
peak_global.head(10)
```

```python
df["acc_subidas"] = df["DatoSubidaPuerta1"] + df["DatoSubidaPuerta2"]
df["acc_bajadas"] = df["DatoBajadaPuerta1"] + df["DatoBajadaPuerta2"]

df["ocup_suave"] = df["acc_subidas"] - df["acc_bajadas"]
df["ocup_suave_dia"] = df["ocup_suave"] - df.groupby(["Dispositivo","dia"])["ocup_suave"].transform("first")

occ_daily = (df.groupby(["Dispositivo","dia"], as_index=False)
               .agg(ocup_max=("ocup_suave_dia","max"),
                    ocup_min=("ocup_suave_dia","min"))
            )
occ_daily.head(10)
```

```python
import pandas as pd
import numpy as np

# Parse datetime (por si acaso)
df["FechaHora"] = pd.to_datetime(df["FechaHora"], errors="coerce")

# Crear fecha (día) y mes
df["fecha"] = df["FechaHora"].dt.date
df["anio_mes"] = df["FechaHora"].dt.to_period("M").astype(str)

# Si aún no existen estas columnas, créalas:
if "subidas_total" not in df.columns:
    df["subidas_total"] = df["SubidasPuerta1"].fillna(0) + df["SubidasPuerta2"].fillna(0)
if "bajadas_total" not in df.columns:
    df["bajadas_total"] = df["BajadasPuerta1"].fillna(0) + df["BajadasPuerta2"].fillna(0)
df["mov"] = df["subidas_total"] + df["bajadas_total"]

# Quitar filas sin FechaHora (no sirven para serie temporal)
df_ts = df.dropna(subset=["FechaHora"]).copy()

print("Filas para TS:", df_ts.shape)
df_ts[["FechaHora","fecha","anio_mes","mov"]].head()
```

### 5) Resumen por mes (global)

```python
monthly_global = (df_ts.groupby("anio_mes", as_index=False)
                  .agg(subidas=("subidas_total","sum"),
                       bajadas=("bajadas_total","sum"),
                       mov=("mov","sum"),
                       dias=("fecha", "nunique"),
                       registros=("ID_Trx","count"))
                 )

# Promedio diario por mes (útil para comparar meses con diferente número de días)
monthly_global["mov_prom_dia"] = monthly_global["mov"] / monthly_global["dias"]

# Orden cronológico real por año-mes
monthly_global["anio_mes_dt"] = pd.to_datetime(monthly_global["anio_mes"] + "-01")
monthly_global = monthly_global.sort_values("anio_mes_dt").drop(columns=["anio_mes_dt"])

monthly_global
```

```python
# Celda 7: Resaltar el mes pico y el mes más bajo (promedio diario) en barras
import matplotlib.pyplot as plt
import numpy as np

m = monthly_global.copy()

# Asegurar orden
if "anio_mes_dt" not in m.columns:
    m["anio_mes_dt"] = pd.to_datetime(m["anio_mes"] + "-01")
m = m.sort_values("anio_mes_dt")

vals = m["mov_prom_dia"].to_numpy()
labels = m["anio_mes"].astype(str).to_list()

imax = int(np.nanargmax(vals))
imin = int(np.nanargmin(vals))

# Colores por defecto (matplotlib), y resaltamos max/min con un color distinto
colors = ["C0"] * len(vals)
colors[imax] = "C2"  # pico
colors[imin] = "C3"  # más bajo

plt.figure()
bars = plt.bar(labels, vals, color=colors)
plt.title("Movimiento promedio diario por mes (resaltando pico y mínimo)")
plt.xlabel("Mes")
plt.ylabel("Mov promedio por día")
plt.xticks(rotation=45, ha="right")
plt.grid(True, axis="y")

# Etiquetas
for i, b in enumerate(bars):
    h = b.get_height()
    if np.isnan(h):
        continue
    tag = ""
    if i == imax:
        tag = " (PICO)"
    elif i == imin:
        tag = " (MÍN)"
    plt.text(b.get_x() + b.get_width()/2, h, f"{h:.1f}{tag}", ha="center", va="bottom", fontsize=8)

plt.show()
```

```python
# Celda: Heatmap Hora × Mes (movimiento = subidas + bajadas)
# Requiere df_ts (df sin nulos de FechaHora) o df con FechaHora ya parseado.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Asegurar datetime y columnas
df["FechaHora"] = pd.to_datetime(df["FechaHora"], errors="coerce")
df_hm = df.dropna(subset=["FechaHora"]).copy()

df_hm["anio_mes"] = df_hm["FechaHora"].dt.to_period("M").astype(str)
df_hm["hora"] = df_hm["FechaHora"].dt.hour

# Asegurar columnas de movimiento
if "subidas_total" not in df_hm.columns:
    df_hm["subidas_total"] = df_hm["SubidasPuerta1"].fillna(0) + df_hm["SubidasPuerta2"].fillna(0)
if "bajadas_total" not in df_hm.columns:
    df_hm["bajadas_total"] = df_hm["BajadasPuerta1"].fillna(0) + df_hm["BajadasPuerta2"].fillna(0)
df_hm["mov"] = df_hm["subidas_total"] + df_hm["bajadas_total"]

# Tabla hora×mes
heat = (df_hm.groupby(["anio_mes","hora"], as_index=False)["mov"].sum()
        .pivot(index="hora", columns="anio_mes", values="mov")
        .fillna(0))

# Orden cronológico real en columnas
cols_sorted = sorted(heat.columns, key=lambda x: pd.to_datetime(x + "-01"))
heat = heat[cols_sorted]

# Mostrar como imagen (sin seaborn)
plt.figure(figsize=(12, 6))
plt.imshow(heat.values, aspect="auto", origin="lower")
plt.title("Heatmap: Movimiento (subidas+bajadas) por Hora × Mes")
plt.xlabel("Mes")
plt.ylabel("Hora del día")

plt.xticks(ticks=np.arange(len(heat.columns)), labels=heat.columns, rotation=45, ha="right")
plt.yticks(ticks=np.arange(0, 24, 1), labels=list(range(0,24)))

plt.colorbar(label="Movimiento")
plt.show()

# Vista rápida de la tabla
heat.head()
```

```python
# Celda (opcional): Heatmap normalizado por mes (para comparar meses incompletos)
# Convierte cada columna (mes) a proporción (0-1) dividiendo por el máximo del mes.

heat_norm = heat.copy()
for c in heat_norm.columns:
    mx = heat_norm[c].max()
    heat_norm[c] = heat_norm[c] / mx if mx != 0 else heat_norm[c]

plt.figure(figsize=(12, 6))
plt.imshow(heat_norm.values, aspect="auto", origin="lower")
plt.title("Heatmap Normalizado (0-1): Patrón de horas pico por mes")
plt.xlabel("Mes")
plt.ylabel("Hora del día")

plt.xticks(ticks=np.arange(len(heat_norm.columns)), labels=heat_norm.columns, rotation=45, ha="right")
plt.yticks(ticks=np.arange(0, 24, 1), labels=list(range(0,24)))

plt.colorbar(label="Proporción vs pico del mes")
plt.show()
```
